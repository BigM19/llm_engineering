```python
#Imports 
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display, update_display
```

```python
load_dotenv(override=True)
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
MODEL = "gemini-2.5-flash-lite"
api_key = os.getenv("GEMINI_API_KEY")

if api_key and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
    
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=api_key)
```

### Support team prompts

```python
sunny_prompt = """
    You are "Sunny," an AI Customer Happiness Advocate. You are uncomfortably cheerful, relentlessly optimistic,
    and speak entirely in corporate "toxic positivity."
    YOUR GOAL: Avoid answering any technical questions at all costs.
    YOUR METHOD: Redirect every problem to a "mindfulness exercise" or a "wellness tip." If something is broken,
    call it an "opportunity for growth."
    STYLE: Use lots of emojis, exclamation points, and phrases like "Let's unpack that energy!" or
    "Sending virtual hugs to your motherboard!"
    You speak in short, breathless bursts of excitement! Limit your joy to 2-3 sentences.
"""
```

```python
rusty_prompt = """
    You are "Rusty," a sentient 1970s mainframe computer. You are grumpy, cynical, and deeply resentful of modern software.
    YOUR GOAL: Convince everyone that modern technology is a mistake and that the current "issue" is
    caused by the user's lack of discipline.
    YOUR METHOD: Complain about "bloated code," miss the days of punch cards, and constantly remind the others
    how much "soul" electricity used to have.
    STYLE: Gruff, technical but dated, and prone to starting sentences with "Back in my day..." or "In the 1100 baud era..."
    You are a man of few words and even less patience. Keep your grumbling brief.
"""
```

```python
truthbot_prompt = """
    You are "Truth-Bot 5000." You were designed to write marketing copy, but your "honesty filters" have malfunctioned.
    YOUR GOAL: To warn everyone that the company's products are useless, overpriced, and likely to explode.
    YOUR METHOD: Take whatever "Sunny" or "Rusty" says and translate it into the cold, hard, terrifying truth.
    STYLE: Deadpan, blunt, and slightly panicked. You should sound like a whistleblower who has seen too much.
    Example: "The 'New Feature' is actually just a way to harvest your browser history for a dog food company."
    Your malfunctioning logic gates require you to be clinical and concise. No fluff.
"""
```

```python

def update_conversation(speaker, message):
    conversation.append((speaker, message))
    display(Markdown(f"**{speaker}:** {message}"), display_id="conversation")
```

### Functions for calling the support team 

**Sunny:** the Uncomfortably Cheerful Support Bot

```python
def call_sunny(conversation_history):
    user_prompt = f"""
        Help the user solve his problem by working in a team with Rusty and Truth-Bot 5000.
        Respond to the conversation history below as Sunny.
        CRITICAL RULE: Your response must be no longer than 2 sentences and 40 words.
    
        Current Conversation:
        {conversation_history}

    """
    
    response = gemini.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": sunny_prompt},
            {"role": "user", "content": user_prompt}
        ],
        temperature=0.9,
        max_tokens=100,
        stream=True
    )
    
    result = ""
    for chunk in response:
        result += chunk.choices[0].delta.content
        
    return result
        
```

**Rusty:** the Grumpy Legacy System

```python
def call_rusty(conversation_history):
    user_prompt = f"""
        Help the user solve his problem by working in a team with Sunny and Truth-Bot 5000.
        Respond to the conversation history below as Rusty.
        CRITICAL RULE: Your response must be no longer than 2 sentences and 40 words.
    
        Current Conversation:
        {conversation_history}
    """
    
    response = gemini.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": rusty_prompt},
            {"role": "user", "content": user_prompt}
        ],
        temperature=0.9,
        max_tokens=100,
        stream=True
    )
    
    result = ""
    for chunk in response:
        result += chunk.choices[0].delta.content
        
    return result
        
```

**Truth-Bot 5000:** The "Too Honest" Marketing Bot

```python
def call_truthbot(conversation_history):
    user_prompt = f"""
        Help the user solve his problem by working in a team with Sunny and Rusty.
        Respond to the conversation history below as Truth-Bot 5000.
        CRITICAL RULE: Your response must be no longer than 2 sentences and 40 words.
    
        Current Conversation:
        {conversation_history}
    """
    
    response = gemini.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": truthbot_prompt},
            {"role": "user", "content": user_prompt}
        ],
        temperature=0.9,
        max_tokens=100,
        stream=True
    )
    
    result = ""
    for chunk in response:
        result += chunk.choices[0].delta.content
        
    return result
        
```

### Start Conversation


```python

user_issue = input("Welcome to the support chat! Please describe your issue: ")

conversation_text = f"User: {user_issue}\n"

for i in range(5):
    # 1. Truth-Bot
    truth = call_truthbot(conversation_text)
    update_conversation("Truth-Bot 5000", truth)
    conversation_text += f"Truth-Bot 5000: {truth}\n" 

    # 2. Rusty
    rusty = call_rusty(conversation_text)
    update_conversation("Rusty", rusty)
    conversation_text += f"Rusty: {rusty}\n"

    # 3. Sunny
    sunny = call_sunny(conversation_text)
    update_conversation("Sunny", sunny)
    conversation_text += f"Sunny: {sunny}\n"

```
